## Task 1 Preform lot-sizing MILP (capacitated, with backorders)

```python
'''
Usage:
    pip install pulp pandas openpyxl
    python task1_lot_sizing.py <demand_workbook.xlsx> [sheet_name]
'''
# import sys # This is only need if using a py file
import pandas as pd
import pulp

# ---------------- Parameters ----------------
K = 1000   # fixed setup cost per batch ($)
h = 15     # holding cost per preform per day ($)
c = 10     # variable production cost per preform ($)
b = 100    # backorder cost per preform per day ($)
M = 100    # maximum batch size (units)
I0 = 20    # initial finished-preform inventory (day 0)
B0 = 0     # initial backorders
T = list(range(1, 21))
```

```python
def load_demand(path, sheet=None):
    df = pd.read_excel(path, sheet_name=sheet if sheet else "Confirmed_20_Day_Demand")
    col = next(c_ for c_ in df.columns if "demand" in str(c_).lower())
    d = df[col].dropna().astype(float).tolist()[:20]
    if len(d) != 20:
        raise ValueError(f"Expected 20 demand values, found {len(d)}")
    return {t: d[t - 1] for t in T}
```

```python
def solve(d):
    m = pulp.LpProblem("Task1_Preform_LotSizing", pulp.LpMinimize)
    x = pulp.LpVariable.dicts("x", T, lowBound=0, cat="Integer")   # production
    y = pulp.LpVariable.dicts("y", T, cat="Binary")                # setup
    I = pulp.LpVariable.dicts("I", [0] + T, lowBound=0)            # end-of-day inventory
    B = pulp.LpVariable.dicts("B", [0] + T, lowBound=0)            # end-of-day backorders

    setup = pulp.lpSum(K * y[t] for t in T)
    var = pulp.lpSum(c * x[t] for t in T)
    hold = pulp.lpSum(h * I[t] for t in T)
    back = pulp.lpSum(b * B[t] for t in T)
    m += setup + var + hold + back

    m += I[0] == I0
    m += B[0] == B0
    for t in T:
        m += I[t - 1] - B[t - 1] + x[t] - d[t] == I[t] - B[t], f"balance_{t}"
        m += x[t] <= M * y[t], f"capacity_{t}"
    m += B[20] == 0, "all_orders_met"

    m.solve(pulp.PULP_CBC_CMD(msg=False))
    status = pulp.LpStatus[m.status]

    plan = pd.DataFrame({
        "day": T,
        "demand": [d[t] for t in T],
        "setup": [int(round(y[t].value())) for t in T],
        "production": [int(round(x[t].value())) for t in T],
        "end_inventory": [round(I[t].value(), 2) for t in T],
        "end_backorder": [round(B[t].value(), 2) for t in T],
    })
    costs = {
        "setup": setup.value(), "variable": var.value(),
        "holding": hold.value(), "backorder": back.value(),
    }
    costs["total"] = sum(costs.values())
    return status, plan, costs

```

```python

path = "ZORA_synthetic_demand_data.xlsx"
sheet = None

demand = load_demand(path, sheet)
status, plan, costs = solve(demand)

print("Status:", status)
print(plan.to_string(index=False))
```

```python
print("\nCost breakdown ($):")
for k_, v in costs.items():
    print(f"  {k_:<10}{v:>12,.2f}")
print("\nSetup days:", plan.loc[plan.setup == 1, "day"].tolist())
# plan.to_csv("task1_plan.csv", index=False)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(9, 4.2))
w = 0.4
ax.bar(plan.day - w / 2, plan.demand, w, label="Demand", color="#9aa7b8")
ax.bar(plan.day + w / 2, plan.production, w, label="Production", color="#2f5d8a")
ax.plot(plan.day, plan.end_inventory, "o-", color="#d9822b", label="End-of-day inventory")
ax.axhline(M, ls="--", lw=0.8, color="grey")
ax.text(20.4, M + 2, "Batch cap", ha="right", va="bottom", fontsize=9, color="grey")
ax.set_xticks(T)
ax.set_xlabel("Planning day")
ax.set_ylabel("Preforms")
ax.set_title("Task 1 optimal plan: production, demand and inventory")
ax.legend(frameon=False, ncol=3, loc="upper left", fontsize=9)
ax.set_ylim(0, 115)
fig.tight_layout()
fig.savefig("task1_plan.png", dpi=200)
```

## Task 2

```python
# Resource consumption tracking from the production plan
quartz_per_preform = 0.08   # tonnes per preform
helium_per_preform = 2.0    # cubic meters per preform

quartz_start = 5.0          # tonnes at day 0
helium_start = 160.0        # m3 at day 0

resource_plan = plan.copy()
resource_plan["quartz_used_t"] = resource_plan["production"] * quartz_per_preform
resource_plan["helium_used_m3"] = resource_plan["production"] * helium_per_preform

resource_plan["cum_quartz_used_t"] = resource_plan["quartz_used_t"].cumsum()
resource_plan["cum_helium_used_m3"] = resource_plan["helium_used_m3"].cumsum()

resource_table = resource_plan[
    ["day", "production", "quartz_used_t", "helium_used_m3",
     "cum_quartz_used_t", "cum_helium_used_m3",
     ]
].copy()

print(resource_table.to_string(index=False))
```

```python
# Input-supply optimization, with production fixed by the Task 1 plan
days = T
production = plan.set_index("day")["production"].to_dict()

model = pulp.LpProblem("Input_Supply_Optimization", pulp.LpMinimize)

# Constant daily quartz delivery; helium order H[t] arrives the following day.
quartz_daily = pulp.LpVariable("quartz_daily_tonnes", lowBound=0)
helium_order = pulp.LpVariable.dicts("helium_order_m3", range(20), lowBound=0)

quartz_inventory = pulp.LpVariable.dicts("quartz_inventory_t", [0] + days, lowBound=0)
helium_inventory = pulp.LpVariable.dicts("helium_inventory_m3", [0] + days, lowBound=0)

model += quartz_inventory[0] == quartz_start
model += helium_inventory[0] == helium_start

for t in days:
    model += (
        quartz_inventory[t]
        == quartz_inventory[t - 1] + quartz_daily
        - quartz_per_preform * production[t],
        f"quartz_balance_{t}",
    )
    # Order placed on day t-1 arrives on day t; H[0] is the day-0 initial order.
    model += (
        helium_inventory[t]
        == helium_inventory[t - 1] + helium_order[t - 1]
        - helium_per_preform * production[t],
        f"helium_balance_{t}",
    )

quartz_procurement_cost = 350 * len(days) * quartz_daily
helium_procurement_cost = 500 * pulp.lpSum(helium_order[t] for t in range(20))
input_holding_cost = pulp.lpSum(
    0.50 * quartz_inventory[t] + helium_inventory[t] for t in days
)
conversion_cost = pulp.lpSum(
    (3.5 + 5.0) * production[t] for t in days
)

model += (
    quartz_procurement_cost
    + helium_procurement_cost
    + input_holding_cost
    + conversion_cost
)

model.solve(pulp.PULP_CBC_CMD(msg=False))
if pulp.LpStatus[model.status] != "Optimal":
    raise RuntimeError(f"Input-supply model status: {pulp.LpStatus[model.status]}")

input_plan = pd.DataFrame({
    "day": days,
    "production": [production[t] for t in days],
    "quartz_delivery_t": [quartz_daily.value()] * len(days),
    "quartz_end_inventory_t": [quartz_inventory[t].value() for t in days],
    "helium_order_placed_m3": [helium_order[t - 1].value() for t in days],
    "helium_end_inventory_m3": [helium_inventory[t].value() for t in days],
})

input_costs = {
    "quartz procurement": pulp.value(quartz_procurement_cost),
    "helium procurement": pulp.value(helium_procurement_cost),
    "input holding": pulp.value(input_holding_cost),
    "conversion": pulp.value(conversion_cost),
}
input_costs["total"] = sum(input_costs.values())

print("Status:", pulp.LpStatus[model.status])
print(f"Constant daily quartz delivery: {quartz_daily.value():.3f} tonnes")
print(input_plan.to_string(index=False))
print("\nInput-supply cost breakdown ($):")
for name, value in input_costs.items():
    print(f"  {name:<22} {value:,.2f}")
```

```python
# 1. Preforms: actual demand vs. production plan
fig_preforms, ax_preforms = plt.subplots(figsize=(10, 4))
ax_preforms.plot(plan["day"], plan["demand"], "o-", label="Actual demand")
ax_preforms.plot(plan["day"], plan["production"], "s-", label="Production plan")
ax_preforms.set(
    xlabel="Day",
    ylabel="Preforms",
    title="Preforms: actual demand vs. production plan",
)
ax_preforms.set_xticks(T)
ax_preforms.grid(True, alpha=0.3)
ax_preforms.legend(frameon=False)
fig_preforms.tight_layout()


# 2. Quartz: actual consumption vs. delivery order
fig_quartz, ax_quartz_line = plt.subplots(figsize=(10, 4))
ax_quartz_line.plot(
    resource_plan["day"], resource_plan["quartz_used_t"],
    "o-", label="Actual quartz demand (use)",
)
ax_quartz_line.plot(
    input_plan["day"], input_plan["quartz_delivery_t"],
    "s-", label="Quartz delivery order",
)
ax_quartz_line.set(
    xlabel="Day",
    ylabel="Quartz (tonnes)",
    title="Quartz: actual demand vs. delivery order",
)
ax_quartz_line.set_xticks(T)
ax_quartz_line.grid(True, alpha=0.3)
ax_quartz_line.legend(frameon=False)
fig_quartz.tight_layout()


# 3. Helium: actual consumption vs. orders arriving that day
fig_helium, ax_helium_line = plt.subplots(figsize=(10, 4))
ax_helium_line.plot(
    resource_plan["day"], resource_plan["helium_used_m3"],
    "o-", label="Actual helium demand (use on day t)",
)
ax_helium_line.plot(
    input_plan["day"], input_plan["helium_order_placed_m3"],
    "s-", label="Order placed day t−1; arrives for day t",
)
ax_helium_line.set(
    xlabel="Arrival / usage day (t)",
    ylabel="Helium (m³)",
    title="Helium: actual demand vs. next-day order arrivals",
)
ax_helium_line.set_xticks(T)
ax_helium_line.grid(True, alpha=0.3)
ax_helium_line.legend(frameon=False)
fig_helium.tight_layout()

plt.show()
```
